```python
import time, random, json, requests
import pandas as pd
from datetime import datetime
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException

def load_job_ids_from_api():
    session = requests.session()
    session.headers.update({
        "User-Agent": "Mozilla/5.0",
        "Accept": "application/vnd.linkedin.normalized+json+2.1",
        "Csrf-Token": csrf_token,
        "Cookie": cookies.strip(),
        "Referer": referer,
        "X-Restli-Protocol-Version": "2.0.0",
    })
    response = session.get(url)
    job_data = []

    if response.status_code == 200:
        data = response.json()
        for item in data["included"][1:]:
            urn = item.get("entityUrn", "")
            if urn.startswith("urn:li:fsd_jobPosting:"):
                job_data.append({
                    "job_id": urn.split(":")[-1],
                    "job_title": item["title"]
                })

    with open(
        f"jobID-{ISorCS_Jobs}-{keyword_name}-7_days_v5.json",
        "w", encoding="utf-8"
    ) as f:
        json.dump(job_data, f, ensure_ascii=False, indent=4)

def scrape_job_descriptions():
    print(f"Scraping dimulai: {datetime.now().strftime('%H:%M:%S')}")
    start = time.perf_counter()

    with open(f"jobID-{ISorCS_Jobs}-{keyword_name}-7_days_v5.json", "r", encoding="utf-8") as f:
        job_list = json.load(f)

    with open("cookies/cookies.json", "r", encoding="utf-8") as f:
        cookies = json.load(f)

    job_data, scraped = [], 0

    for i, job in enumerate(job_list):
        job_url = f"https://www.linkedin.com/jobs/view/{job['job_id']}/"
        driver = webdriver.Chrome()
        driver.get(job_url)

        for cookie in cookies:
            if "sameSite" in cookie and cookie["sameSite"] not in ["Strict", "Lax", "None"]:
                cookie["sameSite"] = "None"
            driver.add_cookie(cookie)
        driver.refresh()
        time.sleep(random.uniform(7, 10))

        try:
            company = driver.find_element(By.CLASS_NAME, "job-details-jobs-unified-top-card__company-name").text
            html = driver.find_element(By.CLASS_NAME, "jobs-description-content__text--stretch").find_element(By.CSS_SELECTOR, "p[dir='ltr']").get_attribute("innerHTML")
            soup = BeautifulSoup(html, "html.parser")
            ul, ol = soup.find_all("ul"), soup.find_all("ol")
            job_desc = ""

            if ul:
                for u in ul:
                    job_desc += "\n".join([li.get_text(strip=True) for li in u.find_all("li")]) + "\n"
            elif ol:
                for o in ol:
                    job_desc += "\n".join([li.get_text(strip=True) for li in o.find_all("li")]) + "\n"
            else:
                job_desc = soup.get_text(separator='\n', strip=True)

            job_data.append({
                "job_title": job["job_title"],
                "company": company,
                "job_description": job_desc
            })
            scraped += 1
            print(f"Job ke-{i+1} selesai")

        except NoSuchElementException:
            print(f"Job ke-{i+1} ({job['job_title']}) mungkin sudah dihapus.")
        finally:
            driver.quit()

    df = pd.DataFrame(job_data)
    output_path = f"{ISorCS_Jobs}_{keyword_name.lower()}_{job_portal}_7_days_v5.xlsx"
    df.to_excel(output_path, index=False, engine="openpyxl")

    end = time.perf_counter()
    print(f"Scraping berakhir: {datetime.now().strftime('%H:%M:%S')}")
    print(f"Waktu scraping: {end - start:.2f} detik | Total scraped: {scraped}/{len(job_list)}")

if __name__ == "__main__":
    load_job_ids_from_api()
    scrape_job_descriptions()
```
